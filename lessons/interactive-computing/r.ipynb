### Make a small table

A **tibble** is the tidyverse's friendly kind of data frame (a table with labelled rows and columns). It is the same idea as a pandas DataFrame, with nicer printing. `suppressPackageStartupMessages` hides the loading messages. This table holds five samples, their condition, and the millions of sequence reads each produced.

````r
suppressPackageStartupMessages({
  library(tibble)
  library(dplyr)
})

df <- tibble(
  sample_id      = c("S01", "S02", "S03", "S04", "S05"),
  condition      = c("control", "control", "treated", "treated", "treated"),
  reads_millions = c(42.1, 39.8, 44.6, 41.0, 38.2)
)
df
````

**What you should see:** a table with columns `sample_id`, `condition` and `reads_millions`, and five rows `S01` to `S05` (42.1, 39.8, 44.6, 41.0, 38.2).

### Average the reads for each condition

`group_by` splits the table by `condition`. `summarise` averages `reads_millions` in each group. The `%>%` symbol (a "pipe") passes the table on to the next step.

````r
# Group-and-summarise — read counts averaged by condition.
df %>%
  group_by(condition) %>%
  summarise(mean_reads = mean(reads_millions))
````

**What you should see:** a table with two rows:

```
  condition mean_reads
1 control   40.95000
2 treated   41.26667
```

### Draw a bar chart

`ggplot2` is R's "grammar of graphics" library. This cell draws the same two averages as a bar chart.

````r
suppressPackageStartupMessages(library(ggplot2))

summary_df <- df %>%
  group_by(condition) %>%
  summarise(mean_reads = mean(reads_millions))

ggplot(summary_df, aes(x = condition, y = mean_reads, fill = condition)) +
  geom_col(width = 0.6) +
  scale_fill_manual(values = c(control = "steelblue", treated = "indianred")) +
  labs(y = "reads (millions)", title = "Mean reads per condition") +
  theme_minimal(base_size = 12) +
  theme(legend.position = "none")
````

**What you should see:** a chart titled "Mean reads per condition", with two bars, blue for `control` and red for `treated`, and the vertical axis labelled "reads (millions)". The bars are nearly the same height, about 41.

### Print your environment

Save the output of this cell. When a notebook stops working months from now, it tells you the versions you were running.

````r
cat("R version:    ", R.version$version.string, "\n")
cat("Platform:     ", R.version$platform, "\n")
cat("Working dir:  ", getwd(), "\n")
conda_env <- Sys.getenv("CONDA_DEFAULT_ENV")
cat("Conda env:    ", ifelse(nzchar(conda_env), conda_env, "(not in conda)"), "\n\n")

for (pkg in c("tibble", "dplyr", "ggplot2", "Seurat")) {
  ver <- tryCatch(
    as.character(packageVersion(pkg)),
    error = function(e) "(not installed)"
  )
  cat(sprintf("%-12s %s\n", pkg, ver))
}
````

**What you should see:** lines like these (your values will differ):

```

Platform:      aarch64-apple-darwin20.0.0
Working dir:   /Users/you/elements-of-style-lessons/lessons/interactive-computing
Conda env:     eos

tibble       3.3.1
dplyr        1.2.1
ggplot2      4.0.3
Seurat       (not installed)
```

`Seurat (not installed)` is normal. A package you have not installed shows `(not installed)`.
