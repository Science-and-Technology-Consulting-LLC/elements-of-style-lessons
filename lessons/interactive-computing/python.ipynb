### Make a small table

`pandas` is the Python library you will meet first for tables. A **DataFrame** is a table with rows and columns that have labels. This one holds five samples, their condition, and the millions of sequence reads each produced.

````python
import pandas as pd

df = pd.DataFrame({
    "sample_id": ["S01", "S02", "S03", "S04", "S05"],
    "condition": ["control", "control", "treated", "treated", "treated"],
    "reads_millions": [42.1, 39.8, 44.6, 41.0, 38.2],
})
df
````

**What you should see:** a table with columns `sample_id`, `condition` and `reads_millions`, and five rows numbered 0 to 4:

```
  sample_id condition  reads_millions
0       S01   control            42.1
1       S02   control            39.8
2       S03   treated            44.6
3       S04   treated            41.0
4       S05   treated            38.2
```

### Average the reads for each condition

`groupby` splits the table by `condition`. `.mean()` averages `reads_millions` in each group.

````python
# A typical "summary by group" — read counts averaged by condition.
df.groupby("condition")["reads_millions"].mean()
````

**What you should see:**

```
condition
control    40.950000
treated    41.266667
Name: reads_millions, dtype: float64
```

### Draw a bar chart

`matplotlib` is the standard plotting library for Python. This cell draws a small bar chart of the same two averages.

````python
import matplotlib.pyplot as plt

summary = df.groupby("condition")["reads_millions"].mean()

fig, ax = plt.subplots(figsize=(4, 3))
summary.plot(kind="bar", ax=ax, color=["steelblue", "indianred"])
ax.set_ylabel("reads (millions)")
ax.set_title("Mean reads per condition")
plt.tight_layout()
plt.show()
````

**What you should see:** a small chart titled "Mean reads per condition", with two bars labelled `control` and `treated`, and the vertical axis labelled "reads (millions)". The bars are nearly the same height, about 41.

### Print your environment

When something stops working six months from now, the first thing you will want to know is "what was I running?" This cell prints the Python version, where Python lives, your operating system, your conda environment, your folder, and the version of four packages. Save its output. Paste it into a bug report.

````python
import sys, platform, os

print("Python:        ", sys.version.split()[0])
print("Executable:    ", sys.executable)
print("Platform:      ", platform.platform())
print("Conda env:     ", os.environ.get("CONDA_DEFAULT_ENV", "(not in conda)"))
print("Working dir:   ", os.getcwd())
print()

for pkg in ["pandas", "numpy", "matplotlib", "scanpy"]:
    try:
        mod = __import__(pkg)
        print(f"{pkg:14s} {getattr(mod, '__version__', '?')}")
    except ImportError:
        print(f"{pkg:14s} (not installed)")
````

**What you should see:** lines like these (your values will differ):


A package you have not installed shows `(not installed)`. You may also see a `FutureWarning` about `__version__`. It is harmless.
